#  For Loop

A `for` loop is used to **repeat a block of code for each item in a sequence**.

A sequence can be:

- A list
- A string
- A tuple
- A range of numbers

### Syntax

```python
for variable in sequence:
    # code to execute

# `in` Keyword in Python

The keyword `in` is used in Python in two important ways:

1. To **iterate through values** in a `for` loop.
2. To **check whether a value exists** inside a collection or sequence.

---

# 1. `in` in a `for` Loop

The keyword `in` tells Python **where to get the values from**.

### Syntax

```python
for variable in sequence:
    # code to execute

````python
A=[0,1,2,3,4,5]#list
B=(0,1,2,3,4,5)#tuple
C={0,1,2,3,4,5}#set
D="hello_World"#string
E={"name":"vishnu","roll_no":"12345"}#dictionary
print(3 in A)
print("h" in D)
print("name" in E )
print("hello" in E)
````

## `for` Loop

A `for` loop is used to **iterate** through the values of a sequence or collection, one value at a time.

### What is Iteration?

**Iteration** means **one complete execution or cycle of a loop**.

````python
for i in A:
    print(i)
````

````python
for i in D:
    print(i)
````

## `for` Loop with a Dictionary

When a `for` loop is used with a dictionary, it **iterates through the keys by default**.
So:

> **`for i in dictionary` → `i` gets each key one by one.**

The values are **not printed by default**.

````python
for i in E:
    print(i)
````

## Getting Values from a Dictionary

If we need to iterate through the **values** of a dictionary, we use the `.values()` method.

> **`for i in dictionary` → iterates through keys**

> **`for i in dictionary.values()` → iterates through values**

````python
for i in E.values():
    print(i)
````

## Using `i` and `j` to Get Keys and Values

When working with a dictionary, we can use **two variables** in a `for` loop to get the **key and value together**.

We use the `.items()` method to get both the key and value.

### Syntax

```python
for key, value in dictionary.items():
    # code
```

Here:

- `key` → stores the **key**
- `value` → stores the **value**
- `.items()` → gives us the **key-value pairs**

````python
for i,j in E.items():
    print(i," ",j)
````

## `range()` in a `for` Loop

The `range()` function is used with a `for` loop when we want to **repeat the loop a specific number of times** or generate a sequence of numbers.

### Syntax

```python
for i in range(start, stop, step):
    # code
```

- `start` → the number where the sequence starts
- `stop` → the number where the sequence stops **(not included)**
- `step` → the amount by which the number increases or decreases

````python
for i in range(2,10,2):
    print(i)
````

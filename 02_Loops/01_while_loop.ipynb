# While Loop

A `while` loop repeatedly executes a block of code as long as the given
condition is `True`.

### Syntax

```python
while condition:
    # code

````python
count = 1
while count < 5:
    print("current value of count is", count)
    count=count+1
````

## Infinite `while` Loop

A `while` loop becomes an infinite loop when the condition never becomes `False`.

This can happen when the variable used in the condition is **not updated** inside the loop.

Here, `count` is always `1` because we never change it.

Therefore:

`count <= 5` → `1 <= 5` → `True`

The condition remains `True` forever, so the loop keeps running.

To stop the loop, we need to update `count`:

```python
count += 1

````python
count = 1
while count < 5:
    print("current value of count is", count)
    
````

## `while` Loop with `else`

The `else` block runs when the `while` loop finishes **normally**, meaning the
condition becomes `False`.

### Syntax

```python
while condition:
    # loop body
else:
    # runs after the loop finishes normally

````python
count = 1
while count < 5:
    print("current value of count is", count)
    count=count+1
else:
    print("*******the count is over ********")    
````

The loop keeps running while `count <= 5`.

When `count` becomes `6`, the condition becomes `False`, so the `while` loop
ends and the `else` block is executed.

````python
num=1
sum=0

while num!=0:
    num=int(input("enter the number to be added"))
    sum=sum+num
    print("latest sum is :",sum)
else:

    print("the loop is over")    
````

###  Nested `while` Loop

A `while` loop can contain another `while` loop.

The inner loop must finish before the outer loop moves to its next iteration.

````python
i = 1

while i <= 3:
    j = 1

    while j <= 2:
        print(f"i = {i}, j = {j}")
        j += 1

    i += 1
````

## ⭐ Star Pattern Problems ⭐

Star pattern problems are simple programs used to practice **loops**, especially
**nested loops**.

In these problems, we print stars (`*`) in different arrangements by controlling
the **number of rows, spaces, and stars**.

They are useful for understanding how nested loops work because:

- The **outer loop** usually controls the rows.
- The **inner loop** usually controls the stars or spaces in each row.
- By changing the loop conditions, we can create different patterns.

#Important Note

The `end=""` prevents Python from moving to a new line after every star.

The `print()` after the inner loop moves the cursor to the next line.

## Role of `i` and `j`

- **`i`** → controls the row
- **`j`** → controls the number of stars in that row

```text
i = 1  →  *
i = 2  →  **
i = 3  →  ***
i = 4  →  ****
i = 5  →  *****

# 1. Increasing Right Triangle

Print a pattern where the number of stars increases by one in every row.

### Expected Output

```text
*
**
***
****
*****

````python
row=5
i=1
while i<=row:
    j=1
    while j<=i:
        print("*",end="")
        j=j+1
        
    print()
    i=i+1       
````

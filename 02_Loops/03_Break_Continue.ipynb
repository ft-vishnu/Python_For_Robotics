## `break` 

The `break` statement is used to **stop the loop completely**.

When Python reaches `break`, the loop **immediately terminates**.

break is used to **control the execution of a loop**.

## Example of `break`

Imagine we are **searching for an item in a list**.

Once we find the item, we **do not need to continue searching** through the rest of the list.

So, we use `break` to **stop the loop** and print `"Item found"`.


````python
A=[0,1,2,3,4,5,6]
key=4
count=0
for i in A:
    count=count+1
    if(i==key):
        print("item found")
        break
    else:
        print("no item found in the list") 
        
print(f"came out of loop,we got the item in {count}-th iteration")           
````

## `continue` Statement

The `continue` statement is used to **skip the current iteration** of a loop and move to the **next iteration**.

Unlike `break`, `continue` **does not stop the loop completely**.

### Example

Suppose we want to print numbers from `1` to `5`, but we want to **skip the number `3`**.



````python
for i in range(1,10):
    if i==3:
        print("____     : this is the 3's place which is skipped")
        continue
    print(i)
    
````

When `i` becomes `3`, Python reaches `continue`.

It **skips the remaining code in that iteration** and goes back to the beginning of the loop for the next value.

> **`continue` → skip the current iteration and continue with the next iteration.**


### `break` vs `continue`

```text
break     → stops the entire loop
continue  → skips only the current iteration
```

## `continue` and Infinite Loop in a `while` Loop

Consider the following code:

```python
i = 1

while i <= 5:
    if i == 3:
        continue

    print(i)
    i = i + 1
```

The problem is that when `i` becomes `3`, the `continue` statement is executed.

```python
if i == 3:
    continue
```

The `continue` statement **skips the remaining code in the current iteration** and goes back to the `while` condition.

Therefore, these statements are skipped:

```python
print(i)
i = i + 1
```

Because `i = i + 1` is skipped, the value of `i` **never changes from `3`**.

### What Happens?

```text
i = 1 → print 1 → i becomes 2
i = 2 → print 2 → i becomes 3
i = 3 → continue → i remains 3
i = 3 → continue → i remains 3
i = 3 → continue → i remains 3
...
```

The loop becomes **infinite** because:

```python
i <= 5
```

is always `True` when `i = 3`.

> **The loop becomes non-terminating because `continue` skips the statement that updates `i`.**

### Correct Version

```python
i = 1

while i <= 5:
    if i == 3:
        i = i + 1
        continue

    print(i)
    i = i + 1
```

### Output

```text
1
2
4
5
```

> **In a `while` loop, make sure the variable controlling the loop condition is updated even when `continue` is executed.**

## `break` 

The `break` statement is used to **stop the loop completely**.

When Python reaches `break`, the loop **immediately terminates**.

break is used to **control the execution of a loop**.

## Example of `break`

Imagine we are **searching for an item in a list**.

Once we find the item, we **do not need to continue searching** through the rest of the list.

So, we use `break` to **stop the loop** and print `"Item found"`.


```python
A=[0,1,2,3,4,5,6]
key=4
count=0
for i in A:
    count=count+1
    if(i==key):
        print("item found")
        break
    else:
        print("no item found in the list") 
        
print(f"we got the item in {count}-th iteration")           
```

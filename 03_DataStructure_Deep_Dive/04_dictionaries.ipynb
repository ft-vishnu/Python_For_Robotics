# Dictionaries

A **dictionary** is a collection of data stored as **key-value pairs**.

````python
student={"name":"Vishnu","roll":12345}
print(student)
````

Here:

- `"name"` → **key**
- `"Vishnu"` → **value**
- `"name": "Vishnu"` → **key-value pair*

## 1. Accessing Values

Use the key to access its value.


````python
student["name"]
````

````python
student["roll"]
````

You can also use `get()`

## Synatx
Dictionary_name.get("key_name")

````python
student.get("name")
````

## 2. Changing Values

Dictionaries are **mutable**, so values can be changed.


````python
student["name"]="arun"
print(student)
````


## 3. Adding New Items

Assign a value to a new key.


````python
student["age"]=23
student["subject"]="Robotics"
````

````python
print(student)
````

## 4. Removing Items

### `pop()`

Removes an item using its key.

````python
student={"name":"Vishnu","roll":12345}
student.pop("roll")
print(student)
````

### `clear()`

Removes all items.


````python
student={"name":"Vishnu","roll":12345}
print("before clearing")
print(student)

print("After clearing")
student.clear()
print(student)
````

## 5. Dictionary Methods

### `keys()`

Returns all keys.


````python
student={"name":"Vishnu","roll":12345}
print(student.keys())
````

### `values()`

Returns all values.

```python
print(student.values())
```

````python
student={"name":"Vishnu","roll":12345}
print(student.values())
````

### `items()`

Returns key-value pairs.

````python
student={"name":"Vishnu","roll":12345}
print(student.items())
````

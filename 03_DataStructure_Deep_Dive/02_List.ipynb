# Lists in Python

A **list** is a collection used to store **multiple values in a single variable**.

A list can contain multiple elements, and the elements are written inside **square brackets `[]`**, separated by commas.

````python
my_list=[1,2.3,"hello",'V']
print(my_list)
````

```text
numbers → name of the list
[]      → square brackets used to create a list
1, 2.3, 'hello', 'V'... → elements of the list
```

A list can contain different types of data:
So, a Python list can store:

- Integers
- Floats
- Strings
- Boolean values
- Other lists
- And more


### Important Properties of Lists

- **Ordered** → elements have a specific position.
- **Indexed** → each element can be accessed using an index.
- **Mutable** → elements can be changed after the list is created.
- **Allows duplicates** → the same value can appear multiple times.


# Creating a List

A list is created by placing elements inside **square brackets `[]`**.

### Syntax

```python
list_name = [element1, element2, element3]
```

````python
my_list=[1,2.3,"hello",'V']
print(my_list)
````

We can also create a list without any elements.




````python

numbers = []

print(numbers)
````

# Accessing List Elements

Since a list is **ordered and indexed**, we can access individual elements using their **index**.

Python uses **zero-based indexing**, which means the first element has index `0`.


````python
lis=["vishnu",123,"arun"]
print(lis[0])
print(lis[1])
print(lis[-1])
````

````python

numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[2])
print(numbers[4])
````


### Index Positions

```text
Element →  10   20   30   40   50
Index   →   0    1    2    3    4
```

So:

```text
numbers[0] → 10
numbers[1] → 20
numbers[2] → 30
numbers[3] → 40
numbers[4] → 50
```

### Negative Indexing

We can also access elements from the **end of the list** using negative indices.

```text
Element →  10   20   30   40   50
Positive→   0    1    2    3    4
Negative→  -5   -4   -3   -2   -1
```


````python
print(numbers[-1])
````


> **Positive indexing starts from `0` and moves from left to right.**

> **Negative indexing starts from `-1` and moves from right to left.**

# 4. Slicing Lists

**Slicing** is used to access a **range of elements** from a list.

### Syntax

```python
list[start:stop]
```

- `start` → starting index **(included)**
- `stop` → ending index **(not included)**


````python
numbers = [10, 20, 30, 40, 50]
print(numbers[0:3])
````

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
# last index is not included in the slicing
````

```text
Element →  10   20   30   40   50
Index   →   0    1    2    3    4
              ↑         ↑
            start      stop


If we don't specify the start index, Python starts from the beginning.



````python

numbers = [10, 20, 30, 40, 50]

print(numbers[:3])
````


We can also specify a `step`.

### Syntax

```python
list[start:stop:step]
```

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0:4:2])
````


> **Indexing → accesses one element**

> **Slicing → accesses multiple elements from a range**

# 5. Changing List Elements

Lists are **mutable**, which means we can **change their elements after the list has been created**.

We can change an element by using its **index**.

### Syntax

```python
list[index] = new_value
```


````python
lis=[1,2,3,4]
print("list before changing :",lis)
lis[0]=100
print("list after changing :",lis)
````

### Changing Multiple Elements

We can also use slicing to change multiple elements.


````python
numbers = [10, 20, 30, 40, 50]
# last index not includede in slicing
print("before changing",numbers)
numbers[1:3]=[200,300]
print("after changing",numbers)
````

**Lists are mutable, so their elements can be changed using indexing or slicing.**

# 6. Adding Elements to a List

Python provides different methods to **add elements to an existing list**.

The three commonly used methods are:

- `append()`
- `insert()`
- `extend()`



# `append()` Method

The `append()` method is used to **add one element to the end of a list**.

### Syntax

```python
list.append(element)
```


````python
numbers = [10, 20, 30]
print("before append:",numbers)
numbers.append(40)
print("after append:",numbers)
````

# `insert()` Method

The `insert()` method is used to **add an element at a specific position (index) in a list**.

### Syntax

```python
list.insert(index, element)
```

- `index` → the position where the element should be inserted
- `element` → the value that we want to add


````python
numbers = [10, 20, 30, 40]
print("before insert: ",numbers)
numbers.insert(2,100)
print("after insert: ",numbers)
````


Here:

```text
Before → [10, 20, 30, 40]
                 ↑
              index 2

After  → [10, 20, 100, 30, 40]
```

The new element `100` is inserted at **index `2`**.

The existing elements are shifted one position to the right.

```text
Index →   0    1    2    3    4
Before → 10   20   30   40
After  → 10   20  100   30   40
```

> **`insert()` → adds an element at a specific index in a list.**

`extend()` Method

The `extend()` method is used to **add multiple elements to the end of a list**.

It takes the elements from another **iterable** and adds them **one by one** to the existing list.
### Syntax

```python
list.extend(iterable)
```


````python
numbers = [10, 20, 30, 40]
print("before extend: ",numbers)
numbers.extend(["vishnu","arun",123,1.232])
print("after extend: ",numbers)
````


> **`append()` → adds one element**

> **`extend()` → adds elements individually from another iterable**

# 7. Removing Elements from a List

Python provides several ways to remove elements from a list.

### `remove()`

Removes a specific **value** from the list.

```python
numbers = [10, 20, 30, 40]

numbers.remove(30)

print(numbers)
```

Output:

```text
[10, 20, 40]
```

> **`remove()` → removes an element by its value.**

---

### `pop()`

Removes an element using its **index** and returns the removed element.

```python
numbers = [10, 20, 30, 40]

x = numbers.pop(2)

print(x)
print(numbers)
```

Output:

```text
30
[10, 20, 40]
```

> **`pop()` → removes an element by index.**

---

### `del`

`del` is used to delete an element or a range of elements using their **index**.

```python
numbers = [10, 20, 30, 40]

del numbers[1]

print(numbers)
```

Output:

```text
[10, 30, 40]
```

> **`del` → deletes elements using their index or a slice.**

---

### `clear()`

Removes **all elements** from the list.

```python
numbers = [10, 20, 30, 40]

numbers.clear()

print(numbers)
```

Output:

```text
[]
```

> **`clear()` → removes all elements from the list.**


# 8. List Methods

### `index()`

Returns the **index of the first occurrence** of a specified value.

```python
numbers = [10, 20, 30, 20]

print(numbers.index(20))
```

Output:

```text
1
```

> **`index()` → finds the position of an element.**

---

### `count()`

Returns the **number of times** a value occurs in a list.

```python
numbers = [10, 20, 30, 20, 20]

print(numbers.count(20))
```

Output:

```text
3
```

> **`count()` → counts how many times an element occurs.**

---

### `sort()`

Arranges the elements in **ascending order** by default.

```python
numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)
```

Output:

```text
[10, 20, 30, 40]
```

> **`sort()` → arranges the elements in order.**

---

### `reverse()`

Reverses the **order of elements** in the list.

```python
numbers = [10, 20, 30, 40]

numbers.reverse()

print(numbers)
```

Output:

```text
[40, 30, 20, 10]
```

> **`reverse()` → reverses the order of elements.**


# 9. List Operators

### `+` Operator

The `+` operator is used to **combine two lists**.

```python
A = [1, 2, 3]
B = [4, 5, 6]

C = A + B

print(C)
```

Output:

```text
[1, 2, 3, 4, 5, 6]
```

> **`+` → combines two lists.**

---

### `*` Operator

The `*` operator is used to **repeat the elements of a list**.

```python
A = [1, 2, 3]

print(A * 2)
```

Output:

```text
[1, 2, 3, 1, 2, 3]
```

> **`*` → repeats a list a specified number of times.**


# 10. Checking Elements

### `in`

The `in` operator checks whether a value **exists in a list**.

It returns a Boolean value:

```text
True  → element exists
False → element does not exist
```

```python
numbers = [10, 20, 30, 40]

print(20 in numbers)
print(50 in numbers)
```

Output:

```text
True
False
```

> **`in` → checks whether an element exists in the list.**

---

### `not in`

The `not in` operator checks whether a value **does not exist in a list**.

```python
numbers = [10, 20, 30, 40]

print(50 not in numbers)
print(20 not in numbers)
```

Output:

```text
True
False
```

> **`not in` → checks whether an element does not exist in the list.**

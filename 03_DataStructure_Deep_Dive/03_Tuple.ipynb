# Tuples 

A **tuple** is a collection used to store **multiple values in a single variable**.

A tuple is similar to a list, but the main difference is that a tuple is **immutable**, meaning its elements cannot be changed after the tuple is created.

Tuples are written using **parentheses `()`**, with elements separated by commas.


````python
numbers=(1,2,4,"one","two")
print(numbers)
````

 **Tuple → an ordered and immutable collection of elements.**

### Single-Element Tuple

To create a tuple with **only one element**, a comma is required.

````python
number=(100,)
print(number)
````

If you don't add the comma, Python does not consider it a tuple./

````python
tup=(10)
print(tup)
print("the class of tup is:",type(tup))
#the but the class we wanted was tuple lol
````

# 3. Accessing Tuple Elements

Tuples are **ordered and indexed**, so we can access their elements using their index number.


### Positive Indexing

Positive indexing starts from **0**.

````python
numbers=(1,2,4,"one","two")
print(numbers[0])
print(numbers[1])
print(numbers[2])
print(numbers[3])
````

### Negative Indexing

Negative indexing starts from **-1** at the last element.

````python
numbers=(1,2,4,"one","two")
print(numbers[-1])
print(numbers[-2])
print(numbers[-3])
print(numbers[-4])
````

**Remember:** Tuple indexing works the same way as list indexing. The main difference is that tuple elements cannot be changed after creation.

## 5. Tuple Immutability

A tuple is **immutable**, which means that once a tuple is created, its elements **cannot be changed, added, or removed**.

### What a Tuple Cannot Do

- ❌ Cannot change an existing element
- ❌ Cannot add a new element
- ❌ Cannot remove an element
- ❌ Cannot use `append()`
- ❌ Cannot use `insert()`

## 6. What You Can Do With a Tuple

Although tuples cannot be modified, you can still perform many operations on them.

### Access Elements

```python
my_tuple = (10, 20, 30)

print(my_tuple[0])
print(my_tuple[-1])
```

**Output:**
```text
10
30
```

### Slice a Tuple

```python
my_tuple = (10, 20, 30, 40, 50)

print(my_tuple[1:4])
```

**Output:**
```text
(20, 30, 40)
```

### Find the Length

Use `len()` to find the number of elements.

```python
my_tuple = (10, 20, 30, 40)

print(len(my_tuple))
```

**Output:**
```text
4
```

## 7. Tuple Methods

Tuples have two main built-in methods.

### `count()`

`count()` returns how many times a particular value appears in the tuple.

```python
my_tuple = (10, 20, 10, 30, 10)

print(my_tuple.count(10))
```

**Output:**
```text
3
```

### `index()`

`index()` returns the index of the **first occurrence** of a value.

```python
my_tuple = (10, 20, 30, 20)

print(my_tuple.index(20))
```

**Output:**
```text
1
```

## 8. Tuple Operators

### `+` Operator

The `+` operator joins two tuples together.

```python
tuple1 = (10, 20)
tuple2 = (30, 40)

result = tuple1 + tuple2

print(result)
```

**Output:**
```text
(10, 20, 30, 40)
```

### `*` Operator

The `*` operator repeats a tuple.

```python
my_tuple = (1, 2)

print(my_tuple * 3)
```

**Output:**
```text
(1, 2, 1, 2, 1, 2)
```

## 9. Checking Elements in a Tuple

### `in`

Checks whether an element exists in the tuple.

```python
my_tuple = (10, 20, 30)

print(20 in my_tuple)
print(50 in my_tuple)
```

**Output:**
```text
True
False
```

### `not in`

Checks whether an element does **not** exist in the tuple.

```python
my_tuple = (10, 20, 30)

print(50 not in my_tuple)
print(20 not in my_tuple)
```

**Output:**
```text
True
False
```

---

## 10. Looping Through a Tuple

A `for` loop can be used to access each element of a tuple one by one.

```python
my_tuple = (10, 20, 30, 40)

for value in my_tuple:
    print(value)
```

**Output:**
```text
10
20
30
40
```

Each iteration takes the **next element** from the tuple.

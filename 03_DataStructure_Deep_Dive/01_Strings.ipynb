# String Methods in Python

String methods are **built-in methods used to perform different operations on strings**.

A string is a sequence of characters:

```python
name = "vishnu"
```

We can use string methods by using the **dot (`.`) operator** after the string or string variable.

### Syntax

```python
string.method()
```
```text
name → string
.    → dot operator
upper() → string method
```

> **String methods allow us to perform different operations on strings, such as changing their case, replacing characters, splitting text, searching for text, and more.**

# Common string Methods
```text
.upper()       → converts the string to uppercase
.lower()       → converts the string to lowercase
.capitalize()  → capitalizes the first character
.title()       → capitalizes the first character of each word
.strip()       → removes spaces from the beginning and end
.replace()     → replaces part of a string
.split()       → splits a string into a list
.find()        → finds the position of a substring
.count()       → counts occurrences of a substring
.startswith()  → checks whether a string starts with a value
.endswith()    → checks whether a string ends with a value
```

````python
name="vishnu"

a_method=name.upper()
print(a_method)
````

````python
b_method=name.find("i")
print(b_method)
````

# Indexing and Slicing in Strings

Strings are **sequences of characters**, and each character has a position called an **index**.

Python allows us to access individual characters using **indexing** and access a group of characters using **slicing**.



**Indexing** is used to access a **single character** from a string.

### Example

```python
name = "Vishnu"
```

The index positions are:

```text
Character → V   i   s   h   n   u
Index     → 0   1   2   3   4   5
```

Python indexing starts from **0**.

### Syntax

```python
string[index]
```


````python
na="hello"
print("0 =",na[0])
print("1 =",na[1])
print("2 =",na[2])
print("3 =",na[3])



````

### Negative Indexing

Python also supports **negative indexing**, which starts from the end of the string.

```text
Character → V    i    s    h    n    u
Positive  → 0    1    2    3    4    5
Negative  → -6  -5   -4   -3   -2   -1
```

````python
na="hello"
print("last index is :",na[-1])
print("secomd last index is :",na[-2])
print("first index is :",na[-5])
````

> **Indexing → used to access one character from a string.**


# Slicing

**Slicing** is used to access a **part of a string**.

Instead of accessing one character, we can extract a sequence of characters.

### Syntax

```python
string[start:stop]
```

- `start` → starting index **(included)**
- `stop` → ending index **(not included)**


````python
name=="vishnuvenugopal"
print(name[0:6])

````

> **Slicing → used to access a range of characters from a string.**

### Slicing with Step

We can also specify a `step`.

### Syntax

```python
string[start:stop:step]
```


````python
name="vishnuvenugopal"
print(name[1:6:2])
name=="vishnuvenugopal"
print(name[0:6])
print(name[6:])
````

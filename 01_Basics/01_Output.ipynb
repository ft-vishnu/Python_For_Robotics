### .format()
  - `.format()` is a string method used to insert variable values into a string using {} placeholders.
  - It is useful when you have a fixed sentence/template but the variable values can change.
  - `.format()` positional indexes start from 0; therefore, for n values, the indexes range from 0 to n-1.

### Syntax
  - print("Text { } { }".format(value1, value2))  
  
  - print("Text {0} {1}".format(value1, value2)) 

```python
x=1
y=22.33
z='hey rama'
print(">>x is a integer with value:{}.\n>>y is a floating point wih value:{}.\n>>z is a string containing he text z:{}.".format(x,y,z))
#best practice is to put numbering for the {}(place holders)
print("-"*25,".format() with indexing","-"*25)
print(">>The value of x is {0}.\n>>The value of y is {1}.\n>>The value of z is {2}.".format(x, y, z))
```

### f-Strings Method
- f-string means formatted string literal.<br>
- It is the modern and recommended way to insert variables/expressions into strings.<br>
- It was introduced in Python 3.6.<br>
- Put f before the string and variables/expressions inside {}.

### Syntax
- print(f"Text {variable1} {variable2}")



```python
a='apple'
b='ball'
n_a=4
n_b=3
print(f">>a for {a}\n>>b for {b}\n>>no of apple:{n_a}\n>>no of ball:{n_b}")
```

### Format specifier
  - Format specifiers are used to control how a variable/value is represented when printing or formatting a string
  

```python
x = 10
y = 22.33
z = "hello"

print("Integer: %d" % x)
print("Float: %f" % y)
print("String: %s" % z)
```

### Controlling decimal places
#### Syntax
  - print("Value: %.nf" % y)
  - n is the no of decimal places that has to be shown

```python
x=10.34343434
print("value of x in 2 decimal place is :%.2f"%x)
```

### Multiple variables

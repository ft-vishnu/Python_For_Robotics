## input( ) Function
- `input()` is a built-in Python function used to take data from the user through the keyboard.
### Syntax
- variable = input("message")
 

```python
x=input("enter your number")
print(f"the value taken from the user is store in x={x}")
```

```python
print(type(x))
```

- `input()` always gives string as its output 
- so we have to change it manually 
- we use typecasting for this method
  - int()
  - float()
  


```python
y=input("enter a floating point")
print(f"the value of y is {y} and the y belongs type: {type(y)}")
```

```python
#type casting
y_new=int(y)
print(f"value of y is {y_new} and its type after type casting is {type(y_new)} ")
```

- so we use int()/float() function and pass the input fun inside for type casting
- syntax
  - x=int(input("enter the number"))

```python
x=int(input("enter the number"))
print(f"value of x is {x} and type of x is {type(x)}")
```

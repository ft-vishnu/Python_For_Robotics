## 1. `if` Statement

The `if` statement is used to execute a block of code only when a given condition is `True`.

## Syntax

```python
if condition:
    statement1
    statement2

````python
ge = int(input("enter the age:"))

if age >= 18:
    print("You are an adult")
````

## 2. `if-else` Statement

Executes one block if the condition is `True`, otherwise executes another block.

### Syntax

```python
if condition:
    statement1
else:
    statement2
```

````python
age = int(input("enter the age:"))
if age >=18:
    print("u r an adult")
else:
    print("u r not an adult")
    
````

## 3. `if-elif-else` Statement

Used when there are multiple conditions.

### Syntax

```python
if condition1:
    statement1
elif condition2:
    statement2
else:
    statement3
```

````python
mark = int(input("enter the mark to know your grade(out of 50):"))
if mark>=40:
    print("ur grade is 'A+'")
elif mark >=20:   
    print("ur grade is 'A'")
else:
    print("ur grade is 'b'")    
````

## 4. Multiple `elif` Statements

Checks several conditions.

### Syntax

```python
if condition1:
    statement1
elif condition2:
    statement2
elif condition3:
    statement3
else:
    statement4
```

````python
mark = int(input("enter the mark to know your grade(out of 50):"))
if mark>=40:
    print("ur grade is 'A+'")
elif mark >=20:   
    print("ur grade is 'A'")
elif mark >=10:   
    print("ur grade is 'B+'")
else:
    print("ur grade is 'C'")    
````

## 5. Nested `if` Statement

An `if` statement inside another `if`.

### Syntax

```python
if condition1:
    if condition2:
        statement
```


````python
age = int(input("Enter the age: "))

has_id = bool(int(input("If you have ID enter 1, if no ID enter 0: ")))

if age >= 18:
    if has_id:
        print("Entry allowed")
````

## 6. Nested `if-else` Statement

A **nested `if-else` statement** means using an `if-else` statement inside another `if` or `else` block.

It is useful when we need to check a second condition only after the first condition is satisfied.

### Syntax

```python
if condition1:
    if condition2:
        statement1
    else:
        statement2
else:
    statement3
```


````python
age = int(input("Enter the age: "))

has_id = bool(int(input("If you have ID enter 1, if no ID enter 0: ")))

if age >= 18:
    if has_id:
        print("Entry allowed")
        
    else:
        print("id required entry not allowed")
else:
    print("age restriction(age below 18)")           
````

## 7. Short-Hand `if` Statement

A **short-hand `if` statement** is a way of writing a simple `if` statement in a single line.

It is useful when there is only **one statement** to execute.
### Important Point

Short-hand `if` should mainly be used for simple conditions. For complex logic, the normal multi-line `if` statement is easier to read.


### Syntax

```python
if condition: statement
```

````python
age = int(input("enter the age:"))
if age >=18:print("u r an adult")
````

# Indentation

Indentation refers to the spaces at the beginning of a code line. In Python, indentation is very important because Python uses indentation to define a block of code.

Unlike languages such as C, C++, and Java, which use `{ }` to define a block of code, Python uses **indentation**.

---

## Example

```python
age = int(input("Enter the age: "))

has_id = bool(int(input("If you have ID enter 1, if no ID enter 0: ")))

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required, entry not allowed")
else:
    print("Age restriction (age below 18)")
```

Here, the spaces before the `if has_id`, `else`, and `print()` statements are important.

For example:

```python
if age >= 18:
    if has_id:
        print("Entry allowed")
```

The `if has_id:` statement is indented because it belongs to the outer `if age >= 18:` statement.

Similarly, the `print("Entry allowed")` statement is further indented because it belongs to the inner `if has_id:` statement.

### Understanding the Indentation

```text
if age >= 18:                 ← Outer if
    if has_id:                ← Inside outer if
        print("Entry allowed") ← Inside inner if
```

Each level of indentation represents a different code block.

---

## Correct Indentation

```python
if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required, entry not allowed")
else:
    print("Age restriction (age below 18)")
```

The indentation clearly shows which statements belong to which `if` or `else` block.

---

## Wrong Indentation

```python
if age >= 18:
if has_id:
    print("Entry allowed")
```

This will produce an **IndentationError** because the statement inside the outer `if` is not indented.

---

## Another Example of Wrong Indentation

```python
if age >= 18:
    if has_id:
    print("Entry allowed")
```

This is also incorrect because `print()` should be indented inside the `if has_id:` block.

---

## Important Rule

Generally, **four spaces** are used for each level of indentation in Python.

For example:

```python
if age >= 18:                    # Level 0
    if has_id:                   # Level 1
        print("Entry allowed")   # Level 2
```

### Indentation Levels

- **Level 0:** `if age >= 18:`
- **Level 1:** `if has_id:`
- **Level 2:** `print("Entry allowed")`

The indentation must be **consistent** throughout the code block.

> **Remember:** In Python, indentation is not just for readability. It tells Python which statements belong to a particular block of code.

## Why Do We Need String Formatting?

Consider the following code:

```python
x = 5
y = 6

print("multiplication of 5 and 6 is", 5 * 6)
```

### What is the problem?

Here, we have stored the values in variables:

```python
x = 5
y = 6
```

But in the `print()` statement, we are **still writing `5` and `6` directly**:

```python
print("multiplication of 5 and 6 is", 5 * 6)
```

This means the `print()` statement is **not using the variables `x` and `y`**.

### What if the values change?

Suppose:

```python
x = 10
y = 20
```

If we use the same `print()` statement:

```python
print("multiplication of 5 and 6 is", 5 * 6)
```

The output will still be:

```text
multiplication of 5 and 6 is 30
```

Even though `x` and `y` have changed!

### What do we actually want?

We want the **current values of `x` and `y`** to automatically appear in the output.

For example:

```python
x = 10
y = 20
```

We want:

```text
multiplication of 10 and 20 is 200
```

So, instead of writing the values directly, we need a way to **insert the values stored in variables into a string dynamically**.

This is called **string formatting**.

> **String formatting allows us to insert variable values into a string dynamically.**

Python provides different ways to do this:

1. `.format()` method
2. `%` formatting
3. f-strings

We will learn these methods one by one.

## Method 1 `.format()` Method

The `.format()` method is one of the ways to perform **string formatting** in Python.

It allows us to **insert the values of variables into a string** without writing the values directly.

### Syntax

```python
"Text {}".format(value)
```

Here:

- `{}` → **placeholder** where the value will be inserted
- `.format()` → supplies the value for the placeholder

````python
x=7
y=8
print("the product of {0} and {1} is {2}".format(x,y,x*y))
````

> **Remember: `.format()` indexing starts from `0`, and the index must match the position of the value.if not this will lead to error**


````python
x=7
y=8
print("the product of {1} and {2} is {3}".format(x,y,x*y))
````

## Method 2: `%` Formatting

`%` formatting is another method of **string formatting** in Python.

It allows us to insert the values of variables into a string using **format specifiers**.

### Basic Syntax

```python
"Text %specifier" % value
```

The `%` symbol tells Python that a value needs to be inserted into the string.

### Common Format Specifiers

| Specifier | Used for |
|---|---|
| `%d` | Integer |
| `%f` | Floating-point number |
| `%s` | String |


````python
x=23
y="arun"
print("age of %s is %d " %(y,x))
````

## Method 3: f-Strings

**f-strings** are another method of string formatting in Python.

They provide a simple way to **insert the values of variables directly into a string**.

An `f` is written before the string, and variables are placed inside `{}`.

### Syntax

```python
f"Text {variable}"
```


````python
x=7
y=3
print(f"the product of {x} and {y} is {x*y}")
````

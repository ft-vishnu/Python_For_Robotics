## Static Methods

A *static method* is a method inside a class that does *not depend on a particular object or the class itself*.

We create it using `@staticmethod`.

### Basic syntax

```python
class ClassName:

    @staticmethod
    def method_name(parameters):
        # method body
        ...
```

Notice something important:

```python
@staticmethod
def method_name(...):
```

There is *no `self`* and *no `cls`*.

---

### Example

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b
```

We can call it using the class:

```python
result = Calculator.add(10, 20)

print(result)
```

Output:

```text
30
```

The method doesn't need:

```python
self
```

because it isn't working with a particular object.

It also doesn't need:

```python
cls
```

because it isn't working with class-level data.

### Compare all three

```text
Normal method
    ↓
self
    ↓
works with an object


Class method
    ↓
cls
    ↓
works with the class


Static method
    ↓
no self / no cls
    ↓
independent function placed inside the class
```

### Example together

```python
class Student:

    school = "Amrita"

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(self.name)

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

    @staticmethod
    def add(a, b):
        return a + b
```

Here:

```python
s1 = Student("Vishnu")
```

`introduce()` → works with the *object* → `self`

```python
Student.change_school("IISc")
```

`change_school()` → works with the *class* → `cls`

```python
Student.add(10, 20)
```

`add()` → doesn't need object or class data → *no `self` or `cls`*

### 🧠 Easy memory

> `self` → object
> `cls` → class
> neither → static method

And with this, *Stage 2 — Core OOP is complete*.


Yes, but there are two different meanings.

If you mean a *static method*, you normally call it through the class:

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

Calculator.add(10, 20)
```

But you *can also call it through an object*:

```python
c1 = Calculator()

c1.add(10, 20)
```

Both work.

However, if you mean:

```python
add(10, 20)
```

with *no class and no object at all*, then no — not unless `add` is separately defined as a normal function outside the class.

So:

```text
Calculator.add()   → ✅
c1.add()           → ✅
add()              → ❌
```

The important point is that `@staticmethod` removes the automatic `self`/`cls` passing; it does *not* remove the function from the class.


# Concept of Static Method

Think about a class as representing a *thing*.

For example, `Drone` represents an actual drone.

```text
Drone
│
├── Data about the drone
│      ├── battery
│      ├── altitude
│      └── speed
│
└── Actions involving the drone
       ├── fly()
       ├── land()
       └── change_altitude()
```

Those actions need to know *which drone* they're working with.

So they use `self`.

---

Now imagine there is a calculation that is useful *while working with drones*, but the calculation doesn't actually involve any particular drone.

For example:

> "Convert meters per second to kilometers per hour."

That calculation doesn't care:

* which drone it is
* what its battery is
* what its altitude is
* what its speed currently is

It is simply a mathematical operation that happens to be useful in the `Drone` context.

So:

```text
Drone
│
├── drone-specific things
│      ├── battery
│      ├── altitude
│      └── fly()
│             ↑
│            self
│
└── general calculation
       └── m/s → km/h
              ↑
         no self needed
```

That's the *conceptual purpose* of a static method.

### The key idea

Don't think:

> "Static method = a function where I don't want `self`."

Think:

> *"This functionality belongs conceptually with this class, but it doesn't need to know anything about a particular object or the class itself."*

That's why it is kept inside the class and marked `@staticmethod`.

And if the function has *no meaningful conceptual relationship to the class at all*, then you usually wouldn't put it inside the class in the first place.

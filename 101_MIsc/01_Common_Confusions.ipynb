# Accessing Data in Python — Brackets and Quotes

When accessing data in Python, the syntax depends on **what type of data you are working with**.

The most important thing to understand is:

> **`[ ]` is used to access something inside a collection.**

What you put **inside `[ ]`** depends on the type of collection.

---

# 1. Accessing a Variable

A normal variable does **not** need brackets.

````python
name="vishnu"
print(name)
````

Here, `name` is already the variable containing the data.

```python
name
```

means:

> Give me the value stored in `name`.

---


# 2. Accessing a List

A **list** is accessed using an **index**.


````python
li=[1,2,3,5,33,"0ne"]
print(li[0])
print(li[3])
print(li[-1])
````

```python
numbers = [10, 20, 30]

print(numbers[0])
print(numbers[1])
print(numbers[2])
```

### Output

```text
10
20
30The numbers inside `[ ]` are the **indexes**.

```text
numbers = [10, 20, 30]
           ↑   ↑   ↑
           0   1   2
```

So:

```python
numbers[0]
```

means:

> Give me the item at index `0`.

### Rule

```text
List → [index]
```

---
# 3. Accessing a Tuple

A tuple works similarly to a list.

# 4. Accessing a Dictionary

A **dictionary** stores data as **key-value pairs**.





````python
dic={
      "name":"arun",
      "age":23,
      "course_id":1234,
      1:1978
    } 
````


To retrieve the value, use the **key** inside `[ ]`.

1. If the key is a string we have to write it inside a double quotes inside the bracket ["Key_name"]
2. fi the key is a int or flot we can directly write in the square bracket

````python
print(dic["name"])
print(dic[1])
````

# Why Do We Use `*` in `*args` and `**` in `**kwargs`?
The `*` and `**` are Python symbols that tell Python **how to handle the arguments**.

The words `args` and `kwargs` are just conventional names.

---

# 1. Why Does `*args` Have One `*`?

```python
def test(*args):
    print(args)
```

The single `*` tells Python:

> **Collect all extra positional arguments into one tuple.**

For example:

```python
test(10, 20, 30)
```

Python collects them as:

```python
args = (10, 20, 30)
```


So:

```text
*args
  ↓
Collect positional arguments
  ↓
Tuple
```

---


# 2. Why Does `**kwargs` Have Two `*`?

```python
def test(**kwargs):
    print(kwargs)
```

The double `**` tells Python:

> **Collect all extra keyword arguments into one dictionary.**

For example:

```python
test(name="Vishnu", age=23)
```

Python collects them as:

```python
kwargs = {
    "name": "Vishnu",
    "age": 23
}
```

So:

```text
**kwargs
   ↓
Collect keyword arguments
   ↓
Dictionary
```

---


# 3. Why One Star vs Two Stars?

Python uses the number of stars to distinguish the two types of arguments:

```text
*   → positional arguments
**  → keyword arguments
```

---

# 4. Using Both Together

We can use both `*args` and `**kwargs` in the same function.

```python
def test(*args, **kwargs):
    print(args)
    print(kwargs)

test(10, 20, name="Vishnu", age=23)
```

### Output

```text
(10, 20)
{'name': 'Vishnu', 'age': 23}
```

Here:

```text
10, 20
  ↓
*args
  ↓
Tuple


name="Vishnu", age=23
  ↓
**kwargs
  ↓
Dictionary
```

---

# 5. `args` and `kwargs` Are Just Names

The names `args` and `kwargs` are **not special Python keywords**.

The stars are what give them their special behavior.

For example, this also works:

```python
def test(*numbers, **details):
    print(numbers)
    print(details)

test(10, 20, name="Vishnu")
```

Output:

```text
(10, 20)
{'name': 'Vishnu'}
```

Here:

```text
*numbers → positional arguments → tuple

**details → keyword arguments → dictionary
```

---

# 🧠 Easy Memory Trick

> **One star `*` → positional arguments → tuple**

> **Two stars `**` → keyword arguments → dictionary**

```text
*args
  ↓
Positional
  ↓
Tuple


**kwargs
   ↓
Keyword
   ↓
Dictionary
```

### Important

The `*` and `**` are what tell Python **how to collect the arguments**.

`args` and `kwargs` are simply the names we normally give to those collected values.

# 🚁 Q2 — Motor Thrust

A quadrotor needs a total thrust of 20 N.

If thrust is equally distributed among 4 motors:

T_motor = Total thrust / Number of motors

Write a function:

calculate_motor_thrust(total_thrust, number_of_motors)

Test with:
- 20 N, 4 motors
- 30 N, 6 motors
- 15 N, 3 motors

Restrictions:
Handle number_of_motors = 0.

Challenge:
Make the function reusable for different drones.

````python
x=20
y=4

def calculate_motor_thrust(total_thrust, number_of_motors):
    if number_of_motors == 0:
        return None
    return total_thrust / number_of_motors

z = calculate_motor_thrust(x, y)

if z is None:
    print("Calculation not possible")
else:
    print(z)
````

# 🧭 Q3 — Robot Heading

A mobile robot represents heading from 0° to 359°.

If the robot is facing 300° and turns +100°:

new_heading = 300 + 100 = 400°

Since heading must stay below 360°,
the new heading becomes 40°.

Write a Python program to calculate the new heading.

Restrictions:
- Final heading must always be between 0° and 359°.
- Handle positive and negative turns.
- Test different headings and turns.

Challenge:
Handle very large turns such as +720° or -500°.

````python
current_heading=300
input_heading=-1000

heading=current_heading+input_heading

if heading >0:
    while heading >359:
       heading=heading-360
    print(heading)
else:
    while heading <0:
        heading=heading+360
    print(heading)
````

Each value is in meters.

Write a Python program that finds:

1. The closest obstacle
2. Its index

If the closest obstacle is below 2.0 m:

Path is clear

Restrictions:
- Use a loop to examine the scan data.
- Do not manually identify the smallest value.

Challenge:

````python
scan = [4.2, 3.8, 7.1, 2.4, 1.8, 5.6, 8.2]

small=scan[0]

for i in scan:
    if i <small:
        small=i

print("small is :",small)

print("smallest distance index is",scan.index(small))

if small<2:
    print("Obstacle too close")
else:
    print("path is clear")
````


### ⚠️ Error Note — `range()` upper limit

```python
range(-5, 5)
```

does **not** include `5`.

It produces:

```text
-5, -4, -3, -2, -1, 0, 1, 2, 3, 4
```

So when you need a range **including the upper value**, use:

```python
range(-5, 6)
```

because the **end value of `range()` is excluded**.

**Rule to remember:**

> `range(start, stop)` → starts at `start`, stops **before** `stop`.

Example:

```python
range(1, 5)  # 1, 2, 3, 4
```

A robot follows this path:

```python
path = [(0, 0), (1, 0), (2, 0), (3, 0), (3, 1), (3, 2)]
```

The robot should move **one step at a time**.

For this exercise, a valid movement means:

* Horizontal difference = `1`, **OR**
* Vertical difference = `1`

Write a program that checks every consecutive pair of positions.

If every movement is valid:

```text
Path is valid
```

Otherwise:

```text
Invalid movement
```

### Restrictions

* Use a loop.
* Compare consecutive positions.
* Don't manually check every pair.

### Challenge

Print the first position where an invalid movement occurs.






````python
path = [(0, 0), (1, 0), (2, 0), (3, 0), (3, 1), (3, 2)]
for i in range(0,len(path)-1):
    first=path[i]
    second=path[i+1]
    
    if second[0]-first[0]==1 or second[1]-first[1]==1:
        print("path is valid")
    else:
        print("path is invalide")    
````

| Operator | Meaning                       |
| -------- | ----------------------------- |
| `and`    | **Both** must be true         |
| `or`     | **At least one** must be true |
| `not`    | **Reverse** true/false        |


# 📡 Q010 — Sensor Health Check

A robot has three sensors:

```python
sensors = {
    "IMU": 98,
    "GPS": 75,
    "LiDAR": 45
}
```

The number represents sensor health percentage.

Rules:

```text
90–100 → Excellent
70–89  → Good
50–69  → Warning
< 50   → Critical
```

### Task

Write a program that goes through the dictionary and prints:

```text
IMU : Excellent
GPS : Good
LiDAR : Critical
```

### Restrictions

* Use a loop.
* Use `if / elif / else`.
* Work directly with the dictionary.
* Don't manually check each sensor.

### Challenge

Count how many sensors are in **Critical** condition.




````python
sensors = {
    "IMU": 98,
    "GPS": 75,
    "LiDAR": 45
}

count=0
for i in sensors.keys():
    if sensors[i] in range(90,101):
        print(f"{i} : Excellent")
        
    elif sensors[i] in range(70,90):
        print(f"{i} : Good") 
        
    elif sensors[i] in range(50,71):
            print(f"{i} : Warning")     
        
    else:
        print(f"{i} : Critical") 
        count=count+1
        
        
        
print(f"No of sensors in critical={count}")        
        
        
        
              
````

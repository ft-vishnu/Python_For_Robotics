# Q1 — Robot Arm DOF

A robot joint that can move independently contributes **1 degree of freedom (DOF)**.

Given:

joints = ["free", "free", "free", "locked", "free", "locked", "free"]

Write a Python program to count the number of available DOF (`"free"` joints).

Expected output:

Degree of freedom = 5

**Challenge:** Make it work for any number of joints.

```python
joints = ["free", "free", "free","locked", "free", "locked", "free"]

count=0
for i in joints:
    
    if i=="free":
        count=count+1
        
print("Degree of freedom=",count)        
```

## 1. Local Variables

A **local variable** is a variable that is created **inside a function**.

It can normally be accessed **only inside that function**.


````python
def greet():
    name="vishnu"
    print("my name is ",name)
    
    
greet()    
````



Here:

```python
name = "Vishnu"
```

is a **local variable** because it was created inside the `greet()` function.

---

## Accessing a Local Variable Outside the Function

A local variable cannot normally be accessed outside the function where it was created.


````python
def greet():
    name="vishnu"
    print("my name is ",name)
    
    
print(name)    
    
    
greet()  
````

Python gives an error because `name` exists only inside `greet()`.

```text
greet()
   │
   └── name = "Vishnu"
          ↑
      Local variable

## Local Variables in Different Functions

Different functions can have local variables with the **same name**.

````python
def greet():
    print("______________________greet()_____________________________")
    name="vishnu"
    print(f"variable name = name \nvalue inside variable 'name' = {name}")
    
    
    
def greetings():
    print("______________________greeting()__________________________")
    name="arun"
    print(f"variable name = name \nvalue inside variable 'name' = {name}")  
    
    
    
    
greet()
greetings()    
````

The two `name` variables are separate.

```text
greet()
    └── name = vishnu

greeting()
    └── name = arun
```

Changing one does not affect the other.

---


## Summary

| Concept | Meaning |
|---|---|
| Local variable | Variable created inside a function |
| Where can it normally be used? | Inside that function |
| Can another function directly use it? | No |
| Can it be accessed outside the function? | No, normally |

# 2. Global Variables

A **global variable** is a variable created **outside all functions**.

Because it is outside the functions, different functions can access the same global variable.

---


````python
name = "Vishnu"       # Global variable


def greet():

    print("______________________greet()_____________________________")

    print(f"variable name = name")
    print(f"value inside variable 'name' = {name}")


def greetings():

    print("______________________greetings()__________________________")

    print(f"variable name = name")
    print(f"value inside variable 'name' = {name}")


greet()
greetings()

print("______________________outside functions_____________________")
print(f"value inside variable 'name' = {name}")
````

## Why Is `name` a Global Variable?

Look at where we created it:

```python
name = "Vishnu"
```

It is **outside both functions**:

```text
name = "Vishnu"       ← GLOBAL VARIABLE
       │
       ├── greet()
       │
       └── greetings()
```

Therefore, both functions can access the same `name`.

---


> **Local = belongs to one function**

> **Global = can be accessed from different parts of the program**

## 4. `global` Keyword

The `global` keyword is used when we want to **modify a global variable from inside a function**.

---


## Example Without `global`

First, look at this:


````python
count=0
def incremental_counter():
    count=count+1
    print(count)
    
    
    
incremental_counter()    
````



Why?

Python sees:

```python
count = count + 1
```

inside the function and assumes that `count` is a **local variable**.

But we are trying to modify the global `count`.

---

````python
count=0
def incremental_counter():
    count=0
    count=count+1
    print(count)
    
    
print("local count value inside the incremental counter is ")    
incremental_counter()   
print("global count value inside the incremental counter is ")  
print(count)
````

# Using the `global` Keyword

We can tell Python:

> "I want to use the global variable, not create a new local variable."

We do this using `global`.


````python
count=11
def incremental_counter():
    global count
    count=count+1
incremental_counter()    
print("global count value inside the incremental counter is ")  
print(count)    
  
    
````

Now the global variable has been changed.

```text
Before function:
count = 10

        increase_count()
              ↓
       global count
              ↓
       count = count + 1
              ↓
After function:
count = 11
```

````python
speed = 10


def increase_speed():

    global speed

    speed = speed + 5

    print("Inside function:", speed)


increase_speed()

print("Outside function:", speed)
````

The same global variable was modified.

---

# Without `global` vs With `global`

````python
speed=100

def inc_speed():
    speed=10
    print(speed)
````

This creates a **new local variable** called `speed`.

The global variable remains:

```text
speed = 10
```

---



### Important

You generally **do not need `global` just to read a global variable**.

```python
speed = 10


def show_speed():

    print(speed)


show_speed()
```

This works because we are only **reading** the global variable.

`global` becomes important when we want to **assign/change** the global variable inside the function.

---

> **Global variables are accessible from many places, so excessive use can make a program harder to track.**

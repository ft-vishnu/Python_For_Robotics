# Functions

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same code again and again, we can write it **once inside a function** and call it whenever we need it.

---

## 1. Creating a Function

A function is created using the `def` keyword.

### Syntax

```python
def function_name():
    # code to be executed
```

- `def` → tells Python that we are creating a function.
- `function_name` → the name we give to the function.
- `()` → contains parameters, if any.
- `:` → marks the beginning of the function body.
- The indented code is the **function body**.


````python
# function to add 2 numbers
def addition():
    x=input("enter the number")
    y=input("enter the second number")
    z=x+y
    print("The sum of the numbers is ",z)
 
````

The function has now been **created**, but it has not run yet.


## 2. Calling a Function

To execute a function, we need to **call** it using its name followed by parentheses.

````python
#function calling
addition()
````

> **Remember:**
>
> `def` → creates the function  
> `function_name()` → calls the function

---

## 3. Why Use Functions?

Functions help us:

- **Reuse** code
- Avoid writing the same code repeatedly
- Organize programs into smaller tasks
- Make code easier to understand and maintain

For example, in robotics we may create separate functions for:

```python
read_sensor()
calculate_error()
calculate_pid()
send_motor_command()
```

Each function can handle one specific task.

> **Main idea:**  
> **Function = define a task once → call it whenever you need that task.**

## 4. Function Body and Indentation

The **function body** is the block of code that belongs to the function.

Python uses **indentation** to identify which statements are part of the function.


````python
def greet():
    print("hello")
    print("welcom to python")
    
````

````python
greet()
````

````python
def greet():
    print("hello")
    print("function ends here")
print("donne")    
    
````

hen the indentation ends, the function body ends.

> **Remember:**  
> `def` starts the function → `:` starts the function body → indentation tells Python which code belongs to the function.

## 5. Parameters and Arguments

A **parameter** is a variable written inside the function definition. It allows the function to receive data.

An **argument** is the actual value that we pass to the function when calling it.

````python
#fun definition
def greet(name):
    print("hello",name)
    
````

````python
#fundtion call
greet("vishnu")
````

Here:

- `name` → **parameter**
- `"Vishnu"` → **argument**

The value `"Vishnu"` is passed into the parameter `name`.


````python
# squaring of a number
def squa(number):
    print("the squared valueof teh input is :",number*number)
    
squa(int(input("enter the number")))   
````

> **Remember:**  
> **Parameter → variable in the function definition**  
> **Argument → actual value passed when calling the function**

## 6. Multiple Parameters

A function can accept **more than one parameter**.

The parameters are separated by commas `,`.

### Syntax

```python
def function_name(parameter1, parameter2):
    # code
```

````python
def addtwo(a,b):
    print("addition result=",a+b)
    
addtwo(int(input("enter the number 1")),int(input("enter the number 2")))   
````

 `a` and `b` → parameters
`5` and `10` → arguments

Python assigns the arguments according to their position:

```text
a → 5
b → 10

> **Remember:**  
> A function can have multiple parameters, and the arguments are normally passed in the same order as the parameters.

> Parameters = Placeholders

> Arguments = Actual Values

## 7. `return` Statement

The `return` statement is used to **send a result back from a function**.

Instead of printing the result inside the function, we can return it and use it somewhere else.


````python
#Displaying result by printing insie the function

def addtwo(a,b):
    print("addition result=",a+b)
    
addtwo(int(input("enter the number 1")),int(input("enter the number 2")))   
````

````python
#displaying result without printing by using return
def addtwo(a,b):
    return a+b
   
    
result=addtwo(int(input("enter the number 1")),int(input("enter the number 2"))) 
print("the sum=",result) 
````

Here:

- `a + b` → calculated inside the function
- `return a + b` → sends the result back
- `result` → stores the returned value


### `return` Ends the Function

When Python reaches `return`, the function immediately stops.


````python
def addtwo(a,b):
    return a+b
    print("this will not run as the fun ends on return statemnt")
    
result=addtwo(5,10) 


print("sum=",result)   
   
````


> **Easy shortcut:**  
> **`print()` → shows the result**  
> **`return` → sends the result back**

## 8. `print()` vs `return`

Both `print()` and `return` can be used inside a function, but they have **different purposes**.

### `print()`

`print()` **displays** a value on the scree

````python
def addtwo(a,b):
    print("sum=",a+b)
   
   
addtwo(5,10)   
````

The result is displayed, but the function does not send the value back for further use.


### `return`

`return` **sends a value back** from the function.

````python
def addtwo(a,b):
    return a+b
 
result=addtwo(5,10)
print("doublethe resuly=",2*result)     
````

### Easy Shortcut

> **`print()` → Show it**  
> **`return` → Give it back**

For robotics, `return` is especially important because functions often need to **calculate something and send the result to another part of the program**.

```text
Sensor → Function → Calculate → return result → Controller
```

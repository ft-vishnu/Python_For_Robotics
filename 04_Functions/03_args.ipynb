## 6. `*args` — allows a function to accept a variable (unknown/any) number of positional arguments.

`*args` allows a function to accept **any number of positional arguments**.

The arguments are collected inside a **tuple**.

````python
def numbers(*args):
    print(args)
    
    
numbers(1,2,3,4,5,6,7)    
````

ere, `args` contains all the values as a tuple.

### Using `*args`

Since `args` is a tuple, we can loop through it.

````python
def numbers(*args):
    total=0
    for i in args:
        total=total+i
        
    return total    
    
    
    
    
    
y=numbers(10,20,30) 
print(y)
y=numbers(10,20,30,40)
print(y)
````


### Why Use `*args`?

Use `*args` when you don't know beforehand **how many positional arguments** the function will receive.

```python
add_numbers(10, 20)

add_numbers(10, 20, 30)

add_numbers(10, 20, 30, 40, 50)
```

All of these can work with the same function.

> **Easy shortcut:**  
> `*args` → **many positional arguments** → stored as a **tuple**

 `**kwargs` — Variable Number of Keyword Arguments

`**kwargs` allows a function to accept a **variable number of keyword arguments**.

In simple words:

> **`**kwargs` = "I don't know how many keyword arguments I'll receive, so collect all of them."**


````python
def studenyt_id(**kwargs):
    print(kwargs)
    
    
studenyt_id(name="vishnu",age="23",sub_id="RAU51")    
````

````python

studenyt_id(name="vijay",age="20",sub_id="AIM51") 
studenyt_id(name="arun",age="23",sub_id="RAU52") 
````

The arguments are collected into a **dictionary**.
## How It Works

When we call:

studenyt_id(name="arun",age="23",sub_id="RAU52") 

Python collects them as:

```python
{
    "name": "Vishnu",
    "age": 22,
    "sub_id": "RAU52"
}


So:

```text
**kwargs → dictionary
```

---

## Accessing `kwargs`

Because `kwargs` is a dictionary, we can access its values using keys.

````python
def student_id(**kwargs):
    print(kwargs["name"])
    
student_id(name="vishnu",age="23",sub_id="RAU51")    
````

## `*args` vs `**kwargs`

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Accepts | Variable positional arguments | Variable keyword arguments |
| Stored as | Tuple | Dictionary |
| Example | `10, 20, 30` | `name="Vishnu", age=22` |

### Easy Memory Trick

> `*args` → **many positional values → tuple**

> `**kwargs` → **many named values → dictionary**

---

## Summary

```text
**kwargs
   ↓
Variable number of keyword arguments
   ↓
Collected into a dictionary
```

**Remember:**

> `*args` → tuple → positional arguments  
> `**kwargs` → dictionary → keyword arguments

So don't think:

args means positional and kwargs means keyword.

Instead think:

``` 1. * means collect positional arguments.```

``` 2. ** means collect keyword arguments. ```

``` args```  and ```kwargs``` are simply the variable names we conventionally give them

You can replace args and kwargs with any valid variable name.

You can change the names, but keeping args and kwargs is usually better for readability and convention.

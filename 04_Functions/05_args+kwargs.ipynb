# Using `*args` and `**kwargs` Together

We can use `*args` and `**kwargs` **together in the same function**.


````python
def student_info(*positional_argument,**keyword_argument):
    print("postitional arguments are stored in tuples")
    print(positional_argument)
    print("keyword arguments are stored in list")
    print(keyword_argument)
    
````

````python
student_info("vishnu",23,"cs181",name="arun",city="chennai",branch="robotics")
````


## What Happened?


Python separates the arguments based on whether they have a **name**.


### Positional Arguments

('vishnu', 23, 'cs181')

These have no parameter names.

They are collected by:

```python
*args
```

and stored as a **tuple**:

### Keyword Arguments

```python
city="Chennai"
branch="Robotics"
```

These have parameter names.

They are collected by:

```python
**kwargs
```

and stored as a **dictionary**:

```python
{
    "city": "Chennai",
    "branch": "Robotics"
}
```

## Important Order

When defining the function:

```python
def student_info(*args, **kwargs):
```

`*args` comes **before** `**kwargs`.


````python
## if order is nopt followed
def student_info(*positional_argument,**keyword_argument):
    print("postitional arguments are stored in tuples")
    print(positional_argument)
    print("keyword arguments are stored in list")
    print(keyword_argument)
    
````


The correct order is to provide positional arguments first, followed by keyword arguments. If a keyword argument is placed before a positional argument, Python will raise an error .

````python
  
student_info("vishnu",23,city="chennai","cs181",name="arun",branch="robotics")    
````

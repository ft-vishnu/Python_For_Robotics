## 1. Positional Arguments

A **positional argument** is an argument whose value is assigned to a parameter based on its **position or order**.

The first argument is assigned to the first parameter, the second argument to the second parameter, and so on.


````python
def details(name,age,roll):
    print("Name of student:",name)
    print("age of student:",age)
    print("roll of student:",roll)
    
details("vishnu",33,26015)    
    
````

> **Easy shortcut:**  
> **Positional → Position matters**
>
> ```text
> 1st argument → 1st parameter
> 2nd argument → 2nd parameter
> 3rd argument → 3rd parameter
> ```

## 2. Keyword Arguments

A **keyword argument** is an argument passed to a function by explicitly specifying the **parameter name**.

Because we specify the parameter name, the **position of the argument does not matter**.


````python
def details(name,age,roll):
    print("Name of student:",name)
    print("age of student:",age)
    print("roll of student:",roll)
    
details(name="vishnu",age=33,roll=26015)    
    
````

Here, **parameter name matters**, so position does not matter.

> **Easy shortcut:**
> **Positional → position matters**  
> **Keyword → parameter name matters**

## 3. Default Arguments

A **default argument** is a value that is automatically used for a parameter when the caller does not provide a value.

The default value is assigned when defining the function.

### Syntax

```python
def function_name(parameter=default_value):
    # code
```

````python
def greet(name="Vishnu"):
    print("Hello", name)

greet()
greet("Rahul") 
````

When `greet()` is called without an argument, Python uses the default value `"Vishnu"`.

When an argument is provided, it **replaces the default value**.


````python
def speed(v=80):
    print("allowed speed in this region is",v)

print("when the argument is not given ")    
speed() 

print("when a new argument is passed") 
speed(100) 
````

> **Easy shortcut:**  
> **Default argument = backup value**
>
> No value given → use the default  
> Value given → use the provided value

## 4. Positional + Keyword Arguments

A function can receive both **positional arguments** and **keyword arguments** in the same function call

````python
def stu(name,age,roll):
    print("name is :",name)
    print("age is :",age)
    print("roll is :",roll)
    
    
stu("vishnu",roll=26015,age=34)   
````

```text
"Vishnu" → positional argument
age=22   → keyword argument
Roll=26015 → keyword argument
```

### Important Rule

**Positional arguments must come before keyword arguments.**


❌ Incorrect:

````python
def stu(name,age,roll):
    print("name is :",name)
    print("age is :",age)
    print("roll is :",roll)
    
    
stu(roll=26015,"vishnu",age=34) 
````

> **Easy shortcut:**  
> **Positional → first**  
> **Keyword → after positional**
